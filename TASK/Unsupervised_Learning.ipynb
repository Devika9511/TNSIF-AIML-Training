**Unsupervised Learning**

**What is Unsupervised Learning?**

Unsupervised Learning is a type of Machine Learning in which the model learns from unlabeled data.

The dataset contains only input features (X) and no target/output (Y).

The algorithm identifies hidden patterns, similarities, or structures in the data without prior knowledge of the correct answers.

The training data contains:

Input (Features / Independent Variables – X)
No Output (Target Variable – Y)

The goal is to discover meaningful groups, relationships, or patterns in the data.

**Advantages**

No labeled data is required.

Can discover hidden patterns and relationships.

Useful for exploratory data analysis.

Helps in customer segmentation and anomaly detection.

**Disadvantages**
Results are harder to evaluate.

Lower prediction accuracy compared to supervised learning.

Clusters may not always have meaningful interpretations.

Choosing the optimal number of clusters can be difficult.

**Types of Unsupervised Learning**

Clustering

Association

**Clustering**

Clustering groups similar data points into clusters based on their similarities. Objects within the same cluster are more similar to each other than to objects in other clusters.

Examples: Customer Segmentation, Image Segmentation, Document Grouping, Disease Pattern Analysis.

Example: Customer Segmentation

Customer    ID	Age	Annual Income

1	22	25

2	24	28

3	40	80

4	42	85

The algorithm groups customers into different segments based on age and income.

```python
#K-Means Clustering Used for: Dividing data into K predefined clusters.

import pandas as pd
from sklearn.cluster import KMeans
# K-Means Clustering
# Used to group similar data into K clusters.

df=pd.DataFrame({
"Age":[22,24,40,42,25,38],
"Income":[25,28,80,85,30,78]
})
model=KMeans(n_clusters=2,random_state=42)
model.fit(df)
print("Cluster Labels:",model.labels_)
```

```python
#Hierarchical Clustering Used for: Building a hierarchy of clusters without specifying K initially.
import pandas as pd
from sklearn.cluster import AgglomerativeClustering

# Hierarchical Clustering
# Used to build clusters in a hierarchical manner.
df=pd.DataFrame({
"Age":[22,24,40,42,25,38],
"Income":[25,28,80,85,30,78]
})
model=AgglomerativeClustering(n_clusters=2)
labels=model.fit_predict(df)
print("Cluster Labels:",labels)
```

```python
#DBSCAN Used for: Finding clusters of arbitrary shapes and detecting outliers.

import pandas as pd
from sklearn.cluster import DBSCAN

# DBSCAN
# Used for clustering with noise and outlier detection.

df=pd.DataFrame({
"Age":[22,24,40,42,25,38],
"Income":[25,28,80,85,30,78]
})
model=DBSCAN(eps=10,min_samples=2)
labels=model.fit_predict(df)
print("Cluster Labels:",labels)

```

**Association**


Association Learning discovers relationships between items that frequently occur together in a dataset.

Examples: Market Basket Analysis, Product Recommendation, Cross Selling.



```python
#Apriori Algorithm Used for: Finding frequent itemsets and association rules.

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori

# Apriori Algorithm
# Used to find frequent itemsets and association rules.
transactions=[
["Milk","Bread"],
["Milk","Butter"],
["Bread","Butter"],
["Milk","Bread","Butter"]
]
te=TransactionEncoder()
data=te.fit(transactions).transform(transactions)

import pandas as pd
df=pd.DataFrame(data,columns=te.columns_)

frequent_items=apriori(df,min_support=0.5,use_colnames=True)

print(frequent_items)
```

**Dimensionality Reduction**


Dimensionality Reduction is a technique used to reduce the number of input features (dimensions) while preserving the important information in the dataset. It helps improve model performance, reduces computation time, and makes data visualization easier.

PCA (Principal Component Analysis)

Stands for: Principal Component Analysis

How it works:
PCA is a linear dimensionality reduction technique that transforms the original features into a smaller set of new features called principal components, which capture the maximum variance in the data.

Best used for:

Reducing high-dimensional data

Faster model training

Preserving overall data structure

Noise reduction



```python

# PCA
# Used to reduce dimensions while preserving maximum variance.

import pandas as pd
from sklearn.decomposition import PCA
df=pd.DataFrame({
"Math":[80,70,90,60,85],
"Science":[78,68,88,58,82],
"English":[85,75,92,65,87]
})
model=PCA(n_components=2)
result=model.fit_transform(df)
print(result)
```

**t-SNE (t-distributed Stochastic Neighbor Embedding)**

Stands for: t-distributed Stochastic Neighbor Embedding

How it works:
t-SNE is a non-linear dimensionality reduction technique that preserves the similarity between nearby data points while projecting them into a lower-dimensional space.

Best used for:

Data visualization

Cluster identification

Visualizing high-dimensional datasets in 2D or 3D

```python
import pandas as pd
from sklearn.manifold import TSNE
# t-SNE
# Used for visualizing high-dimensional data.
df=pd.DataFrame({
"Math":[80,70,90,60,85],
"Science":[78,68,88,58,82],
"English":[85,75,92,65,87]
})

model=TSNE(n_components=2,random_state=42,perplexity=2)
result=model.fit_transform(df)

print(result)
```

**UMAP (Uniform Manifold Approximation and Projection)**

Stands for: Uniform Manifold Approximation and Projection

How it works:
UMAP is a graph-based non-linear dimensionality reduction technique that preserves both local and global relationships in the data while reducing dimensions.

Best used for:

Large datasets

Faster processing than t-SNE

Data visualization

Maintaining clusters and global structure

```python
!pip install umap-learn
import pandas as pd
import umap

# UMAP
# Used for fast dimensionality reduction and visualization.

df=pd.DataFrame({
"Math":[80,70,90,60,85],
"Science":[78,68,88,58,82],
"English":[85,75,92,65,87]
})

model=umap.UMAP(n_components=2,random_state=42)
result=model.fit_transform(df)

print(result)
```
