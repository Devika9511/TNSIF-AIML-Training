Supervised Learning

**What is Supervised Learning?**

Supervised Learning is a machine learning technique where the model is trained using labeled data, meaning each training example includes both the input and the correct output. The model learns the mapping between the input and output so it can accurately predict the output for new, unseen data.

Each training dataset consists of:

Input (Features / Independent Variables – X)
Output (Target / Dependent Variable – Y)

**Advantages of Supervised Learning**

High prediction accuracy

Easy to evaluate model performance

Works well when labeled data is available

Widely used in real-world applications

**Disadvantages of Supervised Learning**

Requires a large amount of labeled data

Data labeling is time-consuming and expensive

Cannot identify hidden patterns without labeled data

**Types of Supervised Learning **

Supervised Learning has two main types:

Regression

Classification

Regression: Regression is used when the target (output) is a continuous numerical value. The model predicts a numeric output based on input features.

Examples: House price, salary, temperature, rainfall, stock price, sales.

Simple Linear Regression (Regression)


Simple Linear Regression is used to predict a continuous value using one independent variable.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
df = pd.DataFrame({
    "Advertising_Budget": [10, 20, 30, 40, 50],
    "Sales": [25, 40, 55, 70, 85]
})

# Independent Variable
X = df[["Advertising_Budget"]]

# Dependent Variable
y = df["Sales"]

# Create Model
model = LinearRegression()

# Train Model
model.fit(X, y)

# Prediction
prediction = model.predict([[35]])

print("Predicted Sales:", prediction[0])

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
df = pd.DataFrame({
    "StudyHours": [2, 3, 5, 6, 8],
    "Attendance": [70, 75, 80, 90, 95],
    "Assignments": [5, 6, 8, 9, 10],
    "Score": [50, 60, 72, 85, 95]
})

# Features and Target
X = df[["StudyHours", "Attendance", "Assignments"]]
y = df["Score"]

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Predict score for a new student
prediction = model.predict([[4, 85, 7]])

print("Predicted Exam Score:", prediction[0])
```

```python
#Polynomial Regression Used for: Predicting continuous values when the relationship between variables is curved (non-linear).

import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Polynomial Regression
# Used when data follows a CURVED relationship.

df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [25, 35, 50, 70, 95]
})

X = df[["Experience"]]
y = df["Salary"]

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

prediction = model.predict(poly.transform([[6]]))

print("Predicted Salary:", prediction[0])
```

```python
#Decision Tree Regression Used for: Predicting continuous values when data has complex or non-linear relationships.

import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# Decision Tree Regression
# Used for NON-LINEAR regression problems.

df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [25, 35, 50, 70, 95]
})

X = df[["Experience"]]
y = df["Salary"]

model = DecisionTreeRegressor(random_state=42)
model.fit(X, y)

prediction = model.predict([[6]])

print("Predicted Salary:", prediction[0])
```

```python
#Random Forest Regression Used for: Improving prediction accuracy by combining multiple decision trees.

import pandas as pd
from sklearn.ensemble import RandomForestRegressor

# Random Forest Regression
# Used for HIGHER accuracy by combining many Decision Trees.

df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [25, 35, 50, 70, 95]
})

X = df[["Experience"]]
y = df["Salary"]

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

prediction = model.predict([[6]])

print("Predicted Salary:", prediction[0])
```

```python
#Support Vector Regression (SVR) Used for: Predicting continuous values when the dataset is small or has complex patterns.

import pandas as pd
from sklearn.svm import SVR

# Support Vector Regression (SVR)
# Used for SMALL datasets and complex relationships.

df = pd.DataFrame({
    "Experience": [1, 2, 3, 4, 5],
    "Salary": [25, 35, 50, 70, 95]
})

X = df[["Experience"]]
y = df["Salary"]

model = SVR(kernel="rbf")
model.fit(X, y)

prediction = model.predict([[6]])

print("Predicted Salary:", prediction[0])
```

**Classification**

Classification is used when the output (target) is categorical (class labels) instead of numerical values. The model predicts which category or class a new data point belongs to.

Examples: Email (Spam/Not Spam), Disease (Yes/No), Loan (Approved/Rejected), Student (Pass/Fail), Fraud Detection

```python
#Logistic Regression Used for: Binary classification problems (Yes/No, Pass/Fail, Spam/Not Spam).

import pandas as pd
from sklearn.linear_model import LogisticRegression

# Logistic Regression
# Used for BINARY classification problems.

df = pd.DataFrame({
    "StudyHours": [1, 2, 3, 4, 5],
    "Pass": [0, 0, 1, 1, 1]
})

X = df[["StudyHours"]]
y = df["Pass"]

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[3]])

print("Prediction (0=Fail, 1=Pass):", prediction[0])




# Decision Tree Classifier Used for: Classification problems with decision rules and non-linear data.

import pandas as pd
from sklearn.tree import DecisionTreeClassifier

# Decision Tree Classifier
# Used for NON-LINEAR classification problems.

df = pd.DataFrame({
    "Age": [18, 22, 25, 35, 40],
    "Buy": [0, 0, 1, 1, 1]
})

X = df[["Age"]]
y = df["Buy"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

prediction = model.predict([[30]])

print("Prediction (0=No, 1=Yes):", prediction[0])




#Random Forest Classified used for: High-accuracy classification using multiple decision trees.

import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Random Forest Classifier
# Used for HIGHER accuracy by combining multiple Decision Trees.

df = pd.DataFrame({
    "Age": [18, 22, 25, 35, 40],
    "Buy": [0, 0, 1, 1, 1]
})

X = df[["Age"]]
y = df["Buy"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
model.fit(X, y)
prediction = model.predict([[30]])
print("Prediction (0=No, 1=Yes):", prediction[0])
```

```python
#Support Vector Machine (SVM) Used for: Small to medium-sized datasets with clear class boundaries.

import pandas as pd
from sklearn.svm import SVC

# Support Vector Machine (SVM)
# Used for classification with CLEAR class separation.

df = pd.DataFrame({
    "StudyHours": [1, 2, 3, 4, 5],
    "Pass": [0, 0, 1, 1, 1]
})

X = df[["StudyHours"]]
y = df["Pass"]

model = SVC(kernel="linear")
model.fit(X, y)

prediction = model.predict([[3]])
print("Prediction (0=Fail, 1=Pass):", prediction[0])


#K-Nearest Neighbors (KNN) Used for: Classification based on the labels of the nearest data points.

import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# K-Nearest Neighbors (KNN)
# Used to classify data based on the nearest neighbors.

df = pd.DataFrame({
    "Age": [18, 20, 25, 30, 35],
    "Buy": [0, 0, 1, 1, 1]
})

X = df[["Age"]]
y = df["Buy"]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)
prediction = model.predict([[28]])
print("Prediction (0=No, 1=Yes):", prediction[0])



#Naive Bayes Used for: Text classification, spam detection, and probabilistic classification.
import pandas as pd
from sklearn.naive_bayes import GaussianNB

# Naive Bayes
# Used for probabilistic classification such as Spam Detection.
df = pd.DataFrame({
    "WordCount": [5, 10, 15, 20, 25],
    "Spam": [0, 0, 1, 1, 1]
})

X = df[["WordCount"]]
y = df["Spam"]

model = GaussianNB()
model.fit(X, y)

prediction = model.predict([[18]])

print("Prediction (0=Not Spam, 1=Spam):", prediction[0])
```
