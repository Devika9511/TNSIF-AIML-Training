# Convolutional Neural Network (CNN)

## What is CNN?

A Convolutional Neural Network (CNN) is a type of Deep Learning model specifically designed for processing image data. It automatically learns important features such as edges, textures, and shapes from images without manual feature extraction.

### CNN Workflow

Input Image

     ↓
Convolution Layer

     ↓
ReLU Activation

     ↓
Pooling Layer

     ↓
Flatten Layer

     ↓
Fully Connected Layer

     ↓
Output

### Key Features
- Automatic feature extraction
- Parameter sharing
- Local connectivity
- High accuracy for image-related tasks

### Applications
- Image Classification
- Face Recognition
- Medical Image Analysis
- Object Detection
- Self-driving Cars

# CNN Layers

## 1. Convolution Layer
- Applies filters (kernels) to extract image features.
- Detects edges, corners, and textures.

## 2. ReLU Activation
Formula:
ReLU(x) = max(0, x)

- Removes negative values.
- Introduces non-linearity.

## 3. Pooling Layer
Reduces feature map size.

Types:
- Max Pooling
- Average Pooling

Benefits:
- Reduces computation
- Prevents overfitting

## 4. Flatten Layer
Converts 2D feature maps into a 1D vector.

## 5. Fully Connected Layer
Uses extracted features for final classification.

# Convolutional Neural Network (CNN)

## Advantages
- Automatically extracts features from images.
- High accuracy for image classification tasks.
- Reduces the number of parameters through parameter sharing.
- Robust to image translation and scaling.
- Efficient in learning complex visual patterns.

## Disadvantages
- Requires a large amount of labeled data.
- Computationally expensive and requires GPUs.
- Training can be time-consuming.
- Not suitable for sequential or time-series data.
- Difficult to interpret learned features.

## Applications / Examples
- Image Classification
- Face Recognition
- Object Detection
- Medical Image Analysis
- Self-driving Cars
- OCR (Handwritten Digit Recognition)

### Real-Life Examples
- Face Unlock in Smartphones
- Google Photos Image Search
- Detecting Tumors in X-ray/MRI Images
- Traffic Sign Recognition

# Recurrent Neural Network (RNN)

## What is RNN?

A Recurrent Neural Network (RNN) is a Deep Learning model designed to process sequential or time-dependent data. It remembers previous inputs using a hidden state, making it suitable for sequence prediction tasks.

### RNN Workflow

Input x₁ → Hidden State → Output y₁

       ↓
Input x₂ → Hidden State → Output y₂

       ↓
Input x₃ → Hidden State → Output y₃


The hidden state stores information from previous time steps.

# RNN Components

## Input Layer
Receives sequential data.

## Hidden Layer
Stores previous information (memory).

## Output Layer
Produces predictions.

## Limitations
- Vanishing Gradient Problem
- Difficult to learn long sequences
- Slow training

## Improved RNN Models

### LSTM (Long Short-Term Memory)
- Remembers long-term information.
- Solves vanishing gradient problem.

### GRU (Gated Recurrent Unit)
- Similar to LSTM.
- Faster and simpler.

# CNN vs RNN

| Feature | CNN | RNN |
|---------|-----|-----|
| Full Form | Convolutional Neural Network | Recurrent Neural Network |
| Data Type | Images | Sequential Data |
| Memory | No | Yes |
| Main Purpose | Feature Extraction | Sequence Learning |
| Popular Variants | AlexNet, VGG, ResNet | LSTM, GRU |
| Applications | Image Classification, Face Recognition, Medical Imaging | NLP, Chatbots, Speech Recognition, Machine Translation, Time Series Forecasting |

## Summary

### CNN
- Best for image processing.
- Uses Convolution, ReLU, Pooling, Flatten, and Fully Connected layers.

### RNN
- Best for sequential data.
- Uses hidden states to remember previous information.
- LSTM and GRU are advanced RNN models for handling long-term dependencies.

## Conclusion

CNN excels at extracting spatial features from images, while RNN is designed to learn patterns in sequential data. Together, they are two of the most widely used deep learning architectures for solving real-world AI problems.

# Recurrent Neural Network (RNN)

## Advantages
- Designed for sequential and time-series data.
- Remembers previous information using hidden states.
- Handles variable-length sequences.
- Effective for text, speech, and sequence prediction.
- Learns temporal dependencies.

## Disadvantages
- Suffers from the vanishing gradient problem.
- Difficult to remember long sequences.
- Slower training due to sequential processing.
- Computationally expensive for long sequences.
- Lower performance than LSTM and GRU.

## Applications / Examples
- Machine Translation
- Speech Recognition
- Chatbots
- Sentiment Analysis
- Next Word Prediction
- Time Series Forecasting

### Real-Life Examples
- Google Translate
- Siri, Alexa, Google Assistant
- Email Spam Detection
- Stock Price Prediction
- Weather Forecasting
