# Deep Learning Fundamentals

## What is Deep Learning?

Deep Learning (DL) is a subset of Machine Learning that uses **Artificial Neural Networks (ANNs)** with multiple hidden layers to learn complex patterns from data.

Unlike traditional Machine Learning, Deep Learning automatically learns features from raw data without requiring manual feature engineering.

### AI vs ML vs DL

| Artificial Intelligence | Machine Learning | Deep Learning |
|-------------------------|------------------|---------------|
| Makes machines intelligent | Learns from data | Learns using neural networks |
| Broad field | Subset of AI | Subset of ML |
| Rule-based + Learning | Algorithms | Deep Neural Networks |

### Applications
- Image Classification
- Object Detection
- Face Recognition
- Speech Recognition
- NLP (ChatGPT)
- Self-driving Cars
- Medical Diagnosis

# Artificial Neural Network (ANN)

ANN is inspired by the human brain.

## Biological Neuron

Dendrites → Cell Body → Axon → Output

## Artificial Neuron

Inputs (x₁,x₂,...,xₙ)

        │
        ▼

Multiply by Weights (w)

        │
        ▼

Add Bias (b)

        │
        ▼

Activation Function

        │
        ▼
Output (y)

## Components

- Input (x): Features of the data
- Weight (w): Importance of each input
- Bias (b): Additional parameter that shifts the output
- Activation Function: Introduces non-linearity
- Output (y): Final prediction

### Artificial Neuron Formula

y = f(Σ(w × x) + b)

## Activation Functions

### Sigmoid
- Output: 0 to 1
- Used for Binary Classification

### Tanh
- Output: -1 to 1
- Better than Sigmoid for centered data

### ReLU (Rectified Linear Unit)
- Output: max(0,x)
- Most commonly used in hidden layers

### Softmax
- Converts outputs into probabilities.
- Used for Multi-class Classification.

# Forward Propagation & Backpropagation

## Forward Propagation

Forward propagation is the process of passing input data through the neural network to generate predictions.

Steps:
1. Input enters the network.
2. Multiply inputs with weights.
3. Add bias.
4. Apply activation function.
5. Produce output.
6. Calculate loss.

Flow:

Input
   ↓
Weights + Bias
   ↓
Activation Function
   ↓
Prediction
   ↓
Loss

---

## Loss Function

Loss measures the difference between the actual output and predicted output.

Goal:
Minimize the loss.

Examples:
- Mean Squared Error (Regression)
- Cross Entropy Loss (Classification)

---

## Backpropagation

Backpropagation is the process of updating weights to reduce the loss.

Steps:
1. Calculate loss.
2. Compute gradients.
3. Update weights.
4. Repeat for every epoch.

Purpose:
Reduce prediction error.

# Gradient Descent & Training Process

## Gradient Descent

Gradient Descent is an optimization algorithm used to minimize the loss function by updating weights.

Weight Update Formula:

New Weight = Old Weight − Learning Rate × Gradient

where:
- Learning Rate (η): Controls the update step size.
- Gradient: Direction of maximum increase in loss.

### Types of Gradient Descent

1. Batch Gradient Descent
   - Uses the entire dataset.
   - Slower but stable.

2. Stochastic Gradient Descent (SGD)
   - Uses one sample at a time.
   - Faster but more noisy.

3. Mini-Batch Gradient Descent
   - Uses small batches of data.
   - Most commonly used.

---

## Important Terms

**Epoch**
- One complete pass through the entire training dataset.

**Batch Size**
- Number of samples processed before updating weights.

**Learning Rate**
- Controls how much the weights are updated.

## Summary

Deep Learning uses Artificial Neural Networks to learn from data. During training, data flows forward to make predictions (Forward Propagation), the error is measured using a Loss Function, and the network learns by updating weights using Backpropagation and Gradient Descent until the loss is minimized.
