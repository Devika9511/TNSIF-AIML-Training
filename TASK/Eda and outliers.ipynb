**Exploratory Data Analysis (EDA)**


Exploratory Data Analysis (EDA) is the process of analyzing and summarizing a dataset to understand its structure, patterns, relationships, and characteristics before building Machine Learning models.

EDA helps answer questions such as:

What does the data contain?

Are there any patterns or trends?

Which features are important?

Are there missing values or outliers?


**Steps in EDA**

Understand the Business Problem

Understand the Dataset

Perform Descriptive Statistics

Univariate Analysis

Bivariate Analysis

Multivariate Analysis

Detect Outliers

Create Visualizations

Draw Insights

Prepare Data for Machine Learning





```python
import pandas as pd

# Create sample dataset containing height and weight of people
df=pd.DataFrame({
"Height":[5.1,5.5,5.8,6.0,6.2,6.1,5.9,5.7,6.3,7.8],
"Weight":[50,55,60,65,70,68,62,58,72,150]
})

# Display dataset
print(df)
```

```python
import matplotlib.pyplot as plt

# Plot height distribution
plt.hist(df["Height"],bins=5)
plt.title("Height Distribution")
plt.xlabel("Height")
plt.ylabel("Frequency")
plt.show()

# Plot weight distribution
plt.hist(df["Weight"],bins=5)
plt.title("Weight Distribution")
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.show()
```

```python
# Calculate first quartile (25%)
Q1=df["Weight"].quantile(0.25)
# Calculate third quartile (75%)
Q3=df["Weight"].quantile(0.75)
# Calculate Interquartile Range
IQR=Q3-Q1
# Calculate lower and upper limits
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

# Find weight outliers
weight_outliers=df[(df["Weight"]<lower)|(df["Weight"]>upper)]
# Print weight outliers
print("Weight Outliers:")
print(weight_outliers)
```

```python
# Calculate first quartile (25%)
Q1=df["Height"].quantile(0.25)

# Calculate third quartile (75%)
Q3=df["Height"].quantile(0.75)

# Calculate Interquartile Range
IQR=Q3-Q1

# Calculate lower and upper limits
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

# Find height outliers
height_outliers=df[(df["Height"]<lower)|(df["Height"]>upper)]

# Print height outliers
print("Height Outliers:")
print(height_outliers)
```

```python
# Calculate quartiles for weight
Q1=df["Weight"].quantile(0.25)
Q3=df["Weight"].quantile(0.75)

# Calculate IQR
IQR=Q3-Q1

# Calculate limits
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

# Calculate mean of weight column
mean=df["Weight"].mean()

# Replace outliers with mean value
df["Weight_Mean"]=df["Weight"].apply(lambda x:mean if x<lower or x>upper else x)

# Display dataset after replacing outliers
print(df)
```

```python
# Calculate quartiles for weight
Q1=df["Weight"].quantile(0.25)
Q3=df["Weight"].quantile(0.75)

# Calculate IQR
IQR=Q3-Q1

# Calculate limits
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

# Calculate median of weight column
median=df["Weight"].median()

# Replace outliers with median value
df["Weight_Median"]=df["Weight"].apply(lambda x:median if x<lower or x>upper else x)

# Display final dataset
print(df)
```
