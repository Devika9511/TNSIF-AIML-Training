**Feature Engineering**


Feature Engineering is the process of creating, transforming, and selecting features from raw data to improve the performance of Machine Learning models.

Raw data is converted into meaningful features that help the model identify patterns and make better predictions.

Why Feature Engineering is Important

Improves model accuracy

Helps algorithms understand data patterns

Removes unnecessary information

Converts raw data into useful input features

**Types of Feature Engineering**

Creating New Features

Creating new features from existing data.

Example:
From a Date column, we can create:

Year

Month

Day

```python
# Feature Engineering
# Creating new features from date information

import pandas as pd

df=pd.DataFrame({
"Date":["2025-01-10","2025-02-15","2025-03-20"]
})

df["Date"]=pd.to_datetime(df["Date"])

df["Year"]=df["Date"].dt.year
df["Month"]=df["Date"].dt.month
df["Day"]=df["Date"].dt.day

print(df)
```

**Handling Missing Values**

Missing values can affect Machine Learning performance. They can be replaced using mean, median, or mode.

```python
# Feature Engineering
# Replacing missing values with mean

import pandas as pd
df=pd.DataFrame({
"Age":[20,25,None,30,35]
})
df["Age"]=df["Age"].fillna(df["Age"].mean())

print(df)
```

**Encoding Categorical Data**

Machine Learning algorithms cannot directly understand text categories, so categorical values are converted into numerical values.

**Label Encoding**

Converts categories into numbers.

Example:

Male → 1
Female → 0

```python
# Label Encoding
# Converts categorical values into numbers

import pandas as pd
from sklearn.preprocessing import LabelEncoder

df=pd.DataFrame({
"Gender":["Male","Female","Male"]
})

encoder=LabelEncoder()

df["Gender"]=encoder.fit_transform(df["Gender"])

print(df)
```

**One Hot Encoding**

Creates separate columns for each category.

Example:

City:

Delhi
Mumbai
Chennai

Converted into:

Delhi | Mumbai | Chennai

```python
# One Hot Encoding
# Creates separate columns for categories

import pandas as pd
df=pd.DataFrame({
"City":["Delhi","Mumbai","Chennai"]
})
df=pd.get_dummies(df)
print(df)
```

**Feature Selection**

Feature selection means choosing important features and removing unnecessary features.

Benefits:

Reduces complexity

Improves model performance

Reduces training time

```python
# Feature Selection
# Removing unnecessary columns

import pandas as pd
df=pd.DataFrame({
"Age":[20,30,40],
"Salary":[30000,50000,70000],
"ID":[1,2,3]
})
df=df.drop("ID",axis=1)
print(df)
```

**Feature Scaling**


Feature Scaling is the process of converting numerical features into a common range so that Machine Learning algorithms can process them effectively.

Example:

Before Scaling:

Age = 25

Salary = 50000

After Scaling:

Age = 0.25

Salary = 0.50

**Why Feature Scaling is Required**

Prevents features with large values from dominating

Improves model accuracy

Speeds up training

Important for distance-based algorithms

Algorithms that require scaling:

KNN

SVM

Logistic Regression

PCA

**Types of Feature Scaling**

**Standardization (StandardScaler)** : Standardization transforms data so that:

Mean = 0

Standard deviation = 1

Formula:

z = (x - mean) / standard deviation

Used in:

Logistic Regression

SVM

PCA

```python
# Standard Scaling
# Converts data into mean=0 and standard deviation=1

import pandas as pd
from sklearn.preprocessing import StandardScaler

df=pd.DataFrame({
"Age":[20,30,40],
"Salary":[30000,50000,70000]
})
scaler=StandardScaler()

scaled=scaler.fit_transform(df)

print(scaled)
```

**Min-Max Scaling**

Min-Max Scaling converts values into a fixed range, usually between 0 and 1.

Formula:

X'=(X-Min)/(Max-Min)

Used when features need the same range.

```python
# Min-Max Scaling
# Converts values between 0 and 1
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df=pd.DataFrame({
"Age":[20,30,40],
"Salary":[30000,50000,70000]
})
scaler=MinMaxScaler()

scaled=scaler.fit_transform(df)

print(scaled)
```

**Robust Scaling**

Robust Scaling uses:

Median

Interquartile Range (IQR)

It is useful when the dataset contains outliers.

Formula:

X'=(X-Median)/IQR

```python
# Robust Scaling
# Handles datasets with outliers

import pandas as pd
from sklearn.preprocessing import RobustScaler

df=pd.DataFrame({
"Age":[20,30,40,100],
"Salary":[30000,50000,70000,200000]
})

scaler=RobustScaler()

scaled=scaler.fit_transform(df)

print(scaled)
```
