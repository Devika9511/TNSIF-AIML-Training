# 📊 Plotly

## 🔹 Definition

**Plotly** is a Python library used for creating **interactive data visualizations**. Unlike traditional plots, Plotly charts allow users to:

> Zoom into graphs  
> Hover over data points  
> Explore data dynamically  

It is widely used in **Exploratory Data Analysis (EDA)** to understand:

--Patterns  
--Trends  
--Relationships between variables  

---

# Advantages of Plotly

**Creates Interactive Charts**  
Users can zoom, hover, and interact with visualizations.

**Easy to Customize**  
Charts can be modified with different colors, titles, and layouts.

**Supports Multiple Visualizations**  
Provides charts like:
- 📊 Bar Charts
- 📈 Line Charts
- 🔵 Scatter Plots
- 📦 Box Plots
- 📉 Histograms
- 🔥 Heatmaps

**Useful for Dashboards and Data Analysis**  
Helps create attractive dashboards and explore datasets easily.

**Works Well with Pandas DataFrames**  
Directly integrates with Pandas for quick visualization.

---

# Applications of Plotly

> Exploratory Data Analysis (EDA)  
> Machine Learning Data Visualization  
> Business Dashboards  
> Data Reporting  
> Trend Analysis

```python
#Line Chart:Used for: Showing trends over time.

import plotly.express as px
import pandas as pd

# Create sample data
df=pd.DataFrame({
"Month":["Jan","Feb","Mar","Apr","May"],
"Sales":[20,35,40,55,70]
})

# Create line chart
fig=px.line(df,x="Month",y="Sales",title="Monthly Sales")
fig.show()
```

```python
#Bar Chart :Used for: Comparing values between categories.

import plotly.express as px
import pandas as pd

df=pd.DataFrame({
"Product":["Laptop","Mobile","Tablet"],
"Sales":[50,80,40]
})

# Create bar chart
fig=px.bar(df,x="Product",y="Sales",title="Product Sales")
fig.show()
```

```python
#Scatter Plot :Used for: Finding relationships between two variables.

import plotly.express as px
import pandas as pd

df=pd.DataFrame({
"Height":[150,160,170,180,190],
"Weight":[50,60,70,80,90]
})

# Create scatter plot
fig=px.scatter(df,x="Height",y="Weight",title="Height vs Weight")
fig.show()
```

```python
#Histogram : Used for: Understanding data distribution
import plotly.express as px
import pandas as pd

df=pd.DataFrame({
"Age":[20,22,25,30,35,40,45,50]
})

# Create histogram
fig=px.histogram(df,x="Age",title="Age Distribution")

fig.show()
```

```python
# Box Plot : Used for: Detecting outliers and understanding data spread.

import plotly.express as px
import pandas as pd
df=pd.DataFrame({
"Salary":[25000,30000,35000,40000,50000,150000]
})

# Create box plot
fig=px.box(df,y="Salary",title="Salary Outlier Detection")

fig.show()
```

```python
#Pie Chart :Used for: Showing percentage contribution of categories.

import plotly.express as px
import pandas as pd

df=pd.DataFrame({
"Category":["A","B","C"],
"Value":[40,30,30]
})

# Create pie chart
fig=px.pie(df,names="Category",values="Value",title="Category Distribution")

fig.show()
```

```python
# Heatmap (Correlation) : Used for: Finding relationships between numerical features.
import plotly.express as px
import pandas as pd

df=pd.DataFrame({
"Age":[20,30,40],
"Salary":[30000,50000,70000],
"Experience":[1,5,10]
})

# Calculate correlation
corr=df.corr()

# Create heatmap
fig=px.imshow(corr,text_auto=True,title="Correlation Heatmap")

fig.show()
```
